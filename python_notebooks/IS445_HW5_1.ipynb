### Data Preparation

```python
import pandas as pd
import altair as alt
```

```python
alt.data_transformers.disable_max_rows()
url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/ufo-scrubbed-geocoded-time-standardized-00.csv"
```

```python
df = pd.read_csv(url, names=['datetime', 'city', 'state', 'country', 'shape',
                             'duration_seconds', 'duration_hours_min', 'comments',
                             'date_posted', 'latitude', 'longitude'],
                 header=None, skiprows=1)
```

```python
col_names = ['datetime', 'city', 'state', 'country', 'shape', 'duration_sec', 'duration_str', 'comments', 'date_posted', 'lat', 'lon']
df = pd.read_csv(url, names=col_names, header=None)
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
df['year'] = df['datetime'].dt.year

df.head()
```

### Charts

```python
# Chart 1, UFO Shapes Distribution
base1 = alt.Chart(df).mark_bar().encode(
    x=alt.X('count():Q', title='Number of Sightings'),
    y=alt.Y('shape:N', sort='-x', title='UFO Shape'),
    color=alt.value('steelblue')
)

chart1 = base1.properties(
    title="Frequency of UFO Shapes Reported Globally",
    width=600,
    height=400
)

chart1
```

```python
# Chart 2
year_slider = alt.binding_range(min=1950, max=2010, step=1, name='Select Year: ')
select_year = alt.selection_point(fields=['year'], bind=year_slider, value={'year': 2000})

base2 = alt.Chart(df).mark_point(filled=True, opacity=0.5).encode(
    x=alt.X('datetime:T', title='Date of Sighting'),
    y=alt.Y('duration_sec:Q', scale=alt.Scale(type='log'), title='Duration (Seconds, Log Scale)'),
    color=alt.Color('shape:N', legend=alt.Legend(title="UFO Shape")),
    tooltip=['city', 'shape', 'duration_str']
)

chart2 = base2.add_params(
    select_year
).transform_filter(
    select_year
).properties(
    title="UFO Sighting Duration Over Time",
    width=600,
    height=400
)

chart2
```

```python
chart1.save('ufo_shapes.json')
chart2.save('ufo_interactive.json')
```
